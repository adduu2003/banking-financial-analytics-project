## Date Conversion and Range Check

Convert the date columns to datetime format and check the available date range to prepare the data for time-based analysis.

```python
account['date'] = pd.to_datetime(account['date'].astype(str), format='%y%m%d')
loan['date'] = pd.to_datetime(loan['date'].astype(str), format='%y%m%d')
trans['date'] = pd.to_datetime(trans['date'].astype(str), format='%y%m%d')
card['issued'] = pd.to_datetime(card['issued'])

print("ACCOUNT DATE")
print("Start date:", account['date'].min())
print("End date:", account['date'].max())

print("\nLOAN DATE")
print("Start date:", loan['date'].min())
print("End date:", loan['date'].max())

print("\nTRANSACTION DATE")
print("Start date:", trans['date'].min())
print("End date:", trans['date'].max())

print("\nCARD ISSUED")
print("Start date:", card['issued'].min())
print("End date:", card['issued'].max())
```

### Foreign Key Validation

Validate the relationships between related datasets to ensure that referenced records exist in their corresponding parent datasets.

```python
print("Account → District")
print(account['district_id'].isin(district['A1']).value_counts())

print("\nClient → District")
print(client['district_id'].isin(district['A1']).value_counts())

print("\nDisposition → Client")
print(disp['client_id'].isin(client['client_id']).value_counts())

print("\nDisposition → Account")
print(disp['account_id'].isin(account['account_id']).value_counts())

print("\nCard → Disposition")
print(card['disp_id'].isin(disp['disp_id']).value_counts())

print("\nLoan → Account")
print(loan['account_id'].isin(account['account_id']).value_counts())

print("\nOrders → Account")
print(orders['account_id'].isin(account['account_id']).value_counts())

print("\nTransactions → Account")
print(trans['account_id'].isin(account['account_id']).value_counts())
```
